#### Q5. Weighted Mean:- A student's final grade is computed as follows:
> - Assignments: 20% (score = 82)
> - Midterm: 30% (score = 76)
> - Final Exam: 50% (score = 88)

> Calculate the weighted mean.

```python
# Scores
assignment = 82
midterm = 76
final_exam = 88

# Weights (in decimal form)
w1 = 0.20
w2 = 0.30
w3 = 0.50

# Step 1: Calculate weighted marks
assignment_marks = assignment * w1
midterm_marks = midterm * w2
final_marks = final_exam * w3

# Step 2: Calculate weighted mean
weighted_mean = assignment_marks + midterm_marks + final_marks
```

```python
# Display results
print("Assignment Contribution =", assignment_marks)
print("Midterm Contribution =", midterm_marks)
print("Final Exam Contribution =", final_marks)
print("Weighted Mean =", weighted_mean)
```

#### Q7. Median (Even Number of Observations):- The monthly expenses (₹) are: 2200, 2500, 1800, 3100, 2900, 2600, 2400, 2800. Find the median.

```python
# Monthly expenses
expenses = [2200, 2500, 1800, 3100, 2900, 2600, 2400, 2800]

# Step 1: Sort the data
expenses.sort()

print("Sorted Data =", expenses)
```

```python
# Step 2: Find the number of observations
n = len(expenses)

# Step 3: Find the two middle values
middle1 = expenses[n // 2 - 1]
middle2 = expenses[n // 2]

# Step 4: Calculate the median
median = (middle1 + middle2) / 2

print("Median =", median)
```

#### Q8. Mode:- Determine the mode of the following dataset: 12, 15, 18, 15, 20, 18, 15, 22, 18, 15. State whether the distribution is unimodal, bimodal, or multimodal.

```python
# Dataset
data = [12, 15, 18, 15, 20, 18, 15, 22, 18, 15]

# Find frequency of each value
frequency = {}

for i in data:
    frequency[i] = data.count(i)

print("Frequencies:", frequency)

# Find maximum frequency
max_freq = max(frequency.values())

# Find all modes
modes = []

for key, value in frequency.items():
    if value == max_freq:
        modes.append(key)

print("Mode(s):", modes)
```

```python
# Check the type of distribution
if len(modes) == 1:
    print("Distribution: Unimodal")
elif len(modes) == 2:
    print("Distribution: Bimodal")
else:
    print("Distribution: Multimodal")
```

#### Q9. Mean, Median and Mode:- For the following dataset: 15, 16, 18, 18, 19, 20, 20, 21, 24.
> **Calculate**:
> 1. Mean
> 2. Median
> 3. Mode

> Comment on the relationship among the three measures.

```python
# Dataset
data = [15, 16, 18, 18, 19, 20, 20, 21, 24]

# Mean
mean = sum(data) / len(data)

# Median
n = len(data)
median = data[n // 2]

# Mode
mode = max(data, key=data.count)
```

```python

# Display Results
print("Mean =", round(mean, 2))
print("Median =", median)
print("Mode =", mode)
```

```python
# Comment
if mean > median:
    print("Mean > Median")
elif mean < median:
    print("Mean < Median")
else:
    print("Mean = Median")

if median == mode:
    print("Median = Mode")
else:
    print("Median != Mode")
```

#### Q11. Geometric Mean:- The annual growth factors of a company's revenue over four years are: 1.08, 1.12, 0.96, 1.10. Calculate the geometric mean growth factor.

```python
# Growth factors
data = [1.08, 1.12, 0.96, 1.10]

# Step 1: Find product of all values
product = 1

for i in data:
    product = product * i

# Step 2: Calculate Geometric Mean
n = len(data)
gm = product ** (1 / n)
```

```python
# Display results
print("Product =", round(product, 4))
print("Geometric Mean =", round(gm, 4))
```

#### Q16. Range:- The daily production (in units) of a factory over eight days is: 125, 138, 142, 119, 151, 136, 145, 132
> **Calculate**:
> 1. The range.
> 2. The coefficient of range.

```python
# Daily production data
production = [125, 138, 142, 119, 151, 136, 145, 132]

# Step 1: Find maximum and minimum values
maximum = max(production)
minimum = min(production)

# Step 2: Calculate Range
range_value = maximum - minimum

# Step 3: Calculate Coefficient of Range
coefficient = (maximum - minimum) / (maximum + minimum)
```

```python
# Display results
print("Maximum Value =", maximum)
print("Minimum Value =", minimum)
print("Range =", range_value)
print("Coefficient of Range =", round(coefficient, 3))
```

#### Q17. Quartiles and Interquartile Range:- The following are the test scores of 15 students: 42, 45, 48, 50, 53, 55, 57, 60, 62, 64, 66, 68, 70, 72, 75
> **Calculate**:
> 1. First Quartile (Q1)
> 2. Third Quartile (Q3)
> 3. Interquartile Range (IQR)
> 4. Quartile Deviation (Semi-IQR)

```python
# Test scores
scores = [42, 45, 48, 50, 53, 55, 57, 60, 62, 64, 66, 68, 70, 72, 75]

# Step 1: Find number of observations
n = len(scores)

# Step 2: Find positions of Q1 and Q3
Q1_pos = (n + 1) / 4
Q3_pos = 3 * (n + 1) / 4

# Step 3: Get Q1 and Q3 values
Q1 = scores[int(Q1_pos) - 1]
Q3 = scores[int(Q3_pos) - 1]

# Step 4: Calculate IQR
IQR = Q3 - Q1

# Step 5: Calculate Quartile Deviation
QD = IQR / 2
```

```python
# Display results
print("Q1 Position =", Q1_pos)
print("Q3 Position =", Q3_pos)
print("Q1 =", Q1)
print("Q3 =", Q3)
print("Interquartile Range =", IQR)
print("Quartile Deviation =", QD)
```

#### Q18. Population Variance and Standard Deviation:- The number of defective items found during six inspections is: 3, 5, 4, 6, 7, 5

> **Calculate**:
> 1. Population mean
> 2. Population variance
> 3. Population standard deviation

```python
# Number of defective items
data = [3, 5, 4, 6, 7, 5]

# Step 1: Calculate Population Mean
mean = sum(data) / len(data)

# Step 2: Calculate Squared Deviations
sum_sq = 0

for i in data:
    sum_sq = sum_sq + (i - mean) ** 2

# Step 3: Calculate Population Variance
variance = sum_sq / len(data)

# Step 4: Calculate Population Standard Deviation
std_dev = variance ** 0.5
```

```python
# Display Results
print("Population Mean =", round(mean, 2))
print("Population Variance =", round(variance, 2))
print("Population Standard Deviation =", round(std_dev, 2))
```
