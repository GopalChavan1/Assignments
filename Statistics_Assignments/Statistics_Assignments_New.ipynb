#### Q1. Arithmetic Mean:- The weekly sales (in units) of a retail store were: 48, 52, 61, 57, 49, 63, 70
> **Calculate**:
> 1. The arithmetic mean.
> 2. The deviation of each observation from the mean.
> 3. Verify that the sum of deviations from the mean is zero.


```python
sales = [48, 52, 61, 57, 49, 63, 70]       # Weekly sales data
mean = sum(sales) / len(sales)             #Calculate Arithmetic Mean
deviations = [x - mean for x in sales]      #Calculate deviation of each observation
```

```python
print("Mean:", mean)
print("Deviations:", deviations)
print("Sum of deviations:", sum(deviations))
```

#### Q3. Effect of a New Observation:- The mean of eight observations is 27. A ninth observation with value 45 is added. Find the new mean.

```python

mean_8 = 27          # Mean of first 8 observations
n_8 = 8              # Number of observations
new_value = 45       # New observation
total = mean_8 * n_8   # Step 1: Calculate total of first 8 observations
new_total = total + new_value    # Step 2: Add the new observation
new_mean = new_total / (n_8 + 1)   # Step 3: Calculate new mean


```
