Average Purchase Value by Age Group and Gender

```python
import pandas as pd
df = pd.read_csv("customer_data.csv")
bins = [18, 25, 35, 45, 60, 80]
labels = ["18-25", "26-35", "36-45", "46-60", "61-80"]
df["AgeGroup"] = pd.cut(df["Age"],bins=bins,labels=labels,include_lowest=True)
avg_purchase = df.groupby(["AgeGroup", "Gender"], observed=False)["TotalAmountSpent"].mean()
print("Average Purchase Value by Age Group and Gender")
print(avg_purchase)
```

Total Purchase Amount by Location

```python
locpurchase = df.groupby("Location")["TotalAmountSpent"].sum()
print("Total Purchase Amount by Location")
print(locpurchase)
```
