### Exercise Set 1 — Series & DataFrame Basics


### 1. Create a Series of 5 students' marks with custom index (student names).


Explanation:

A Series is a 1-dimensional labeled array in Pandas

Use pd.Series() to create a Series

Pass the index parameter to set custom labels (student names)

This allows easy access to data using student names instead of numeric indices

```python
import pandas as pd

# Create a Series with marks and custom index (student names)
marks = pd.Series(
    [85, 92, 78, 88, 95],  # Data: marks values
    index=['Alice', 'Bob', 'Charlie', 'Diana', 'Ethan'],  # Custom index: student names
    name='Marks'  # Name the Series
)

# Display the Series
print(marks)

# Access specific student's mark using index
print(f"\nBob's marks: {marks['Bob']}")
print(f"Alice's marks: {marks['Alice']}")
```

### 2. Create a DataFrame from a list of dicts representing 4 employees 
   (Name, Age, Dept, Salary). Add a 'Bonus' column = 10% of Salary.

Explanation:

Use pd.DataFrame() with a list of dictionaries to create DataFrame

Each dictionary represents one row (employee)

Add new column by simple assignment: df['Bonus'] = df['Salary'] * 0.10

Pandas performs vectorized operations automatically on entire columns

```python
import pandas as pd

# Create DataFrame from list of dictionaries
employees = pd.DataFrame([
    {'Name': 'John', 'Age': 28, 'Dept': 'DS', 'Salary': 75000},
    {'Name': 'Sarah', 'Age': 35, 'Dept': 'ML', 'Salary': 85000},
    {'Name': 'Mike', 'Age': 32, 'Dept': 'DS', 'Salary': 68000},
    {'Name': 'Emma', 'Age': 29, 'Dept': 'AI', 'Salary': 92000}
])

# Display original DataFrame
print("Original DataFrame:")
print(employees)

# Add Bonus column = 10% of Salary (vectorized operation)
employees['Bonus'] = employees['Salary'] * 0.10

# Display updated DataFrame
print("\nDataFrame with Bonus column:")
print(employees)

# Verify Bonus calculation
print(f"\nJohn's Bonus: {employees['Bonus'][0]} (10% of {employees['Salary'][0]})")
```

### 3. Rename columns: 'Dept' → 'Department', 'Salary' → 'CTC'.


Explanation:

Use df.rename() method with columns parameter to rename specific columns

Pass a dictionary where keys are old names and values are new names

Use inplace=True to modify the original DataFrame, or assign to new variable

Alternative: Use df.columns to rename all columns at once

```python
import pandas as pd

# Create DataFrame (from Q2)
employees = pd.DataFrame([
    {'Name': 'John', 'Age': 28, 'Dept': 'DS', 'Salary': 75000},
    {'Name': 'Sarah', 'Age': 35, 'Dept': 'ML', 'Salary': 85000},
    {'Name': 'Mike', 'Age': 32, 'Dept': 'DS', 'Salary': 68000},
    {'Name': 'Emma', 'Age': 29, 'Dept': 'AI', 'Salary': 92000}
])
employees['Bonus'] = employees['Salary'] * 0.10

# Display before renaming
print("Before renaming:")
print(employees.columns.tolist())

# Rename columns using rename() method
employees = employees.rename(columns={
    'Dept': 'Department',  # Old name → New name
    'Salary': 'CTC'        # Old name → New name
})

# Display after renaming
print("\nAfter renaming:")
print(employees.columns.tolist())
print("\nUpdated DataFrame:")
print(employees)

# Alternative: Rename all columns at once
# employees.columns = ['Name', 'Age', 'Department', 'CTC', 'Bonus']
```

## Exercise Set 2 — Indexing & Selection


### 4. Using .loc, select employees whose Salary > 60000 AND Dept == 'DS'.


Explanation:

.loc[] is used for label-based indexing (can use column names)

Create a boolean mask with conditions: (df['CTC'] > 60000) & (df['Department'] == 'DS')

Use & for AND, | for OR conditions

Pass the mask to .loc[] to filter rows

Note: Use & not and in Pandas boolean operations

```python
import pandas as pd

# Create DataFrame
employees = pd.DataFrame([
    {'Name': 'John', 'Age': 28, 'Dept': 'DS', 'Salary': 75000},
    {'Name': 'Sarah', 'Age': 35, 'Dept': 'ML', 'Salary': 85000},
    {'Name': 'Mike', 'Age': 32, 'Dept': 'DS', 'Salary': 68000},
    {'Name': 'Emma', 'Age': 29, 'Dept': 'AI', 'Salary': 92000}
])
employees['Bonus'] = employees['Salary'] * 0.10
employees = employees.rename(columns={'Dept': 'Department', 'Salary': 'CTC'})

# Display all employees
print("All Employees:")
print(employees)

# Create boolean mask for conditions
# Note: Use & for AND, enclose conditions in parentheses
mask = (employees['CTC'] > 60000) & (employees['Department'] == 'DS')

# Use .loc[] with boolean mask to select rows
filtered_employees = employees.loc[mask]

print("\nEmployees with CTC > 60000 AND Department == 'DS':")
print(filtered_employees)

# Access specific column using .loc
print(f"\nNames of filtered employees: {filtered_employees['Name'].tolist()}")
```

### 5. Using .iloc, select rows 1-3 and columns 0-2.


Explanation:

.iloc[] is used for integer-based indexing (positions, not names)

Syntax: df.iloc[row_start:row_end, col_start:col_end]

Important: Python slicing is exclusive at the end, so 1:4 gets rows 1, 2, 3

Similarly, 0:3 gets columns 0, 1, 2

Returns a new DataFrame with selected rows and columns

```python
import pandas as pd

# Create DataFrame
employees = pd.DataFrame([
    {'Name': 'John', 'Age': 28, 'Dept': 'DS', 'Salary': 75000},
    {'Name': 'Sarah', 'Age': 35, 'Dept': 'ML', 'Salary': 85000},
    {'Name': 'Mike', 'Age': 32, 'Dept': 'DS', 'Salary': 68000},
    {'Name': 'Emma', 'Age': 29, 'Dept': 'AI', 'Salary': 92000}
])
employees['Bonus'] = employees['Salary'] * 0.10

# Display all employees with column positions
print("All Employees (with column positions):")
print(employees)
print(f"\nColumn positions: {list(employees.columns)}")
print("Position 0: Name, Position 1: Age, Position 2: Dept, Position 3: Salary, Position 4: Bonus")

# Select rows 1-3 (use 1:4 for inclusive) and columns 0-2 (use 0:3)
selected = employees.iloc[1:4, 0:3]

print("\nSelected rows 1-3 and columns 0-2 using .iloc:")
print(selected)

# Explanation: 
# - Rows: 1, 2, 3 (Sarah, Mike, Emma)
# - Columns: 0, 1, 2 (Name, Age, Dept)
```

### 6. Set 'Name' as index and access data for a specific employee.


```python
import pandas as pd

# Create DataFrame
employees = pd.DataFrame([
    {'Name': 'John', 'Age': 28, 'Dept': 'DS', 'Salary': 75000},
    {'Name': 'Sarah', 'Age': 35, 'Dept': 'ML', 'Salary': 85000},
    {'Name': 'Mike', 'Age': 32, 'Dept': 'DS', 'Salary': 68000},
    {'Name': 'Emma', 'Age': 29, 'Dept': 'AI', 'Salary': 92000}
])
employees['Bonus'] = employees['Salary'] * 0.10

# Display before setting index
print("Before setting 'Name' as index:")
print(employees)
print(f"\nCurrent index: {employees.index.tolist()}")

# Set 'Name' column as index
employees = employees.set_index('Name')

# Display after setting index
print("\nAfter setting 'Name' as index:")
print(employees)
print(f"\nNew index: {employees.index.tolist()}")

# Access specific employee using .loc with index
print("\n--- Accessing Sarah's data using .loc['Sarah'] ---")
sarah_data = employees.loc['Sarah']
print(sarah_data)

# Access specific column for Sarah
print(f"\nSarah's Age: {employees.loc['Sarah', 'Age']}")
print(f"Sarah's Salary: {employees.loc['Sarah', 'Salary']}")

# Access multiple employees
print("\n--- Accessing John and Mike ---")
print(employees.loc[['John', 'Mike']])
```

## Exercise Set 3 — Data Cleaning


### 7. Given a DataFrame with missing values, fill numeric NaN with column mean,
   and string NaN with 'Unknown'.


Explanation:

Use pd.NA or np.nan to represent missing values

Use df.fillna() for filling missing values

For numeric columns: Calculate mean using df['col'].mean() then fill

For string columns: Use df['col'].fillna('Unknown')

Use df.select_dtypes() to separate numeric and string columns

```python
import pandas as pd
import numpy as np

# Create DataFrame with missing values
df = pd.DataFrame({
    'Name': ['John', 'Sarah', np.nan, 'Emma', np.nan],
    'Age': [28, np.nan, 32, 29, 35],
    'Dept': ['DS', 'ML', 'DS', np.nan, 'AI'],
    'Salary': [75000, 85000, np.nan, 92000, 78000]
})

print("Original DataFrame with missing values:")
print(df)
print(f"\nMissing values per column:")
print(df.isnull().sum())

# Fill numeric columns (Age, Salary) with column mean
for col in df.select_dtypes(include=[np.number]).columns:
    mean_val = df[col].mean()
    df[col] = df[col].fillna(mean_val)
    print(f"Filled '{col}' with mean: {mean_val}")

# Fill string columns (Name, Dept) with 'Unknown'
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna('Unknown')
    print(f"Filled '{col}' with 'Unknown'")

print("\nDataFrame after cleaning:")
print(df)
print(f"\nMissing values after cleaning:")
print(df.isnull().sum())
```

### 8. Find and remove duplicate rows.


Explanation:

Use df.duplicated() to identify duplicate rows (returns boolean Series)

Use df.drop_duplicates() to remove duplicates

Default keeps the first occurrence, can use keep='last' to keep last

Use subset参数 to check duplicates in specific columns only

```python
import pandas as pd

# Create DataFrame with duplicate rows
df = pd.DataFrame({
    'Name': ['John', 'Sarah', 'John', 'Mike', 'Sarah', 'Emma'],
    'Age': [28, 35, 28, 32, 35, 29],
    'Dept': ['DS', 'ML', 'DS', 'DS', 'ML', 'AI'],
    'Salary': [75000, 85000, 75000, 68000, 85000, 92000]
})

print("Original DataFrame with duplicates:")
print(df)

# Identify duplicate rows (returns True for duplicates)
duplicates = df.duplicated()
print("\nDuplicate rows (True = duplicate):")
print(duplicates)

# Show which rows are duplicates
print("\nRows that are duplicates:")
print(df[duplicates])

# Remove duplicates (keeps first occurrence by default)
df_clean = df.drop_duplicates()

print("\nDataFrame after removing duplicates:")
print(df_clean)

# Alternative: Remove duplicates based on specific columns only
df_clean_name_age = df.drop_duplicates(subset=['Name', 'Age'])
print("\nDataFrame after removing duplicates based on Name & Age only:")
print(df_clean_name_age)

# Count duplicates removed
print(f"\nDuplicate rows removed: {len(df) - len(df_clean)}")
```

### 9. Convert an 'Age' column stored as strings to integers.


Explanation:

Use df['col'].astype(int) to convert column type

Use pd.to_numeric() for safer conversion (handles errors)

String numbers like '28' can be directly converted to int

If there are non-numeric strings, use pd.to_numeric(col, errors='coerce') to convert to NaN

```python
import pandas as pd

# Create DataFrame with Age as strings
df = pd.DataFrame({
    'Name': ['John', 'Sarah', 'Mike', 'Emma'],
    'Age': ['28', '35', '32', '29'],  # Stored as strings
    'Dept': ['DS', 'ML', 'DS', 'AI']
})

print("Original DataFrame (Age as strings):")
print(df)
print(f"\nAge column type: {df['Age'].dtype}")

# Method 1: Direct conversion using astype()
df['Age'] = df['Age'].astype(int)
print("\nAfter conversion using astype(int):")
print(df)
print(f"Age column type: {df['Age'].dtype}")

# Method 2: Using pd.to_numeric() (safer for messy data)
df2 = pd.DataFrame({
    'Name': ['John', 'Sarah', 'Mike'],
    'Age': ['28', '35', '32']  # Clean numeric strings
})
df2['Age'] = pd.to_numeric(df2['Age'], errors='coerce')
print("\nUsing pd.to_numeric():")
print(df2)
print(f"Age column type: {df2['Age'].dtype}")

# Handle messy data with non-numeric values
df3 = pd.DataFrame({
    'Name': ['John', 'Sarah', 'Mike'],
    'Age': ['28', '35', 'N/A']  # Contains non-numeric
})
df3['Age'] = pd.to_numeric(df3['Age'], errors='coerce')
print("\nHandling messy data (N/A becomes NaN):")
print(df3)
```

### 10. Use str.strip() and str.title() to clean a messy 'Name' column.


Explanation:

str.strip() removes leading and trailing whitespace

str.title() converts to title case (first letter capital, rest lowercase)

Use df['col'].str.strip() and df['col'].str.title() for string operations

Can chain multiple string methods: df['col'].str.strip().str.title()

```python
import pandas as pd

# Create DataFrame with messy names
df = pd.DataFrame({
    'Name': ['  john  ', 'SARAH', 'mike', '  Emma  ', 'DAVID'],
    'Age': [28, 35, 32, 29, 33]
})

print("Original DataFrame with messy names:")
print(df)

# Step 1: Remove leading/trailing spaces using str.strip()
df['Name'] = df['Name'].str.strip()
print("\nAfter str.strip() (removed spaces):")
print(df)

# Step 2: Convert to title case using str.title()
df['Name'] = df['Name'].str.title()
print("\nAfter str.title() (proper capitalization):")
print(df)

# Alternative: Chain both methods
df2 = pd.DataFrame({
    'Name': ['  john  ', 'SARAH', 'mike'],
    'Age': [28, 35, 32]
})
df2['Name'] = df2['Name'].str.strip().str.title()
print("\nChained operations (strip + title):")
print(df2)

# Other useful string methods
print("\n--- Other string methods ---")
df3 = pd.DataFrame({'Name': ['  john  ', 'SARAH', 'mike']})
print(f"Upper: {df3['Name'].str.upper().tolist()}")
print(f"Lower: {df3['Name'].str.lower().tolist()}")
print(f"Length: {df3['Name'].str.len().tolist()}")
```
