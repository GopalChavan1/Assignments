**Q28.**  
Given `scores = np.array([72, 55, 91, 38, 85, 63, 78, 44])`:
- Use `np.argsort` to get indices that would sort the array ascending
- Use those indices (fancy indexing) to produce the sorted array
- Use `np.argsort` on the result of `np.argsort` to assign ranks (1-based, 1 = smallest)

---

````python
import numpy as np

# Create the array
scores = np.array([72, 55, 91, 38, 85, 63, 78, 44])

# Step 1: Use np.argsort to get indices that would sort the array ascending
sorted_indices = np.argsort(scores)

# Step 2: Use those indices (fancy indexing) to produce the sorted array
sorted_scores = scores[sorted_indices]

# Step 3: Use np.argsort on the result of np.argsort to assign ranks (1-based, 1 = smallest)
# The double argsort gives us the rank positions
rank_positions = np.argsort(sorted_indices)
# Convert to 1-based ranks (1 = smallest)
ranks = rank_positions + 1

# Print all results
print("Original scores:", scores)
print("Sorted indices (from argsort):", sorted_indices)
print("Sorted array (using fancy indexing):", sorted_scores)
print("Rank positions (from double argsort):", rank_positions)
print("1-based ranks:", ranks)

# Verify: show ranks alongside original scores
print("\nVerification - scores with their ranks:")
for i in range(len(scores)):
    print(f"scores[{i}] = {scores[i]}, rank = {ranks[i]}")
````

**Q31.**  
Compute the **Z-score** of `np.array([10, 20, 30, 40, 50])` manually:
- Subtract the mean
- Divide by the standard deviation

Use `np.mean` and `np.std`. Verify the Z-scored array has mean ≈ 0 and std ≈ 1.

---

````python
import numpy as np

# Create the array
data = np.array([10, 20, 30, 40, 50])

# Step 1: Calculate the mean using np.mean
mean_value = np.mean(data)

# Step 2: Calculate the standard deviation using np.std
std_value = np.std(data)

# Step 3: Compute Z-score manually: (x - mean) / std
z_scored_array = (data - mean_value) / std_value

# Step 4: Verify the Z-scored array has mean ≈ 0 and std ≈ 1
z_score_mean = np.mean(z_scored_array)
z_score_std = np.std(z_scored_array)

# Print all results
print("Original data:", data)
print("Mean:", mean_value)
print("Standard deviation:", std_value)
print("Z-scored array:", z_scored_array)
print(f"Mean of Z-scored array: {z_score_mean}")
print(f"Std of Z-scored array: {z_score_std}")

# Verify with formatted output
print("\nVerification:")
print(f"  Mean ≈ 0: {abs(z_score_mean) < 0.0001}")
print(f"  Std ≈ 1: {abs(z_score_std - 1) < 0.0001}")
````


````python
import numpy as np

# Given data
y_true = np.array([1, 0, 1, 1, 0, 0, 1, 0])
y_pred = np.array([0.9, 0.2, 0.8, 0.7, 0.3, 0.4, 0.85, 0.1])

print("True Labels (y_true):")
print(y_true)

print("\nPredicted Probabilities (y_pred):")
print(y_pred)

# 1. Clip predictions to avoid log(0)
y_pred_clipped = np.clip(y_pred, 1e-7, 1 - 1e-7)

print("\nClipped Predictions:")
print(y_pred_clipped)

# 2. Compute Binary Cross-Entropy Loss
loss_terms = (
    y_true * np.log(y_pred_clipped)
    + (1 - y_true) * np.log(1 - y_pred_clipped)
)

print("\nLoss Terms:")
print(loss_terms)

# 3. Compute Mean BCE Loss
bce_loss = -np.mean(loss_terms)

print("\nBinary Cross-Entropy Loss:")
print(bce_loss)

# 4. Verification
print("\n=== VERIFICATION ===")

print("Mean of Loss Terms :", np.mean(loss_terms))
print("Final BCE Loss     :", -np.mean(loss_terms))

# Individual contribution of each sample
individual_losses = -loss_terms

print("\nIndividual Sample Losses:")
print(individual_losses)

print("\nAverage of Individual Losses:")
print(np.mean(individual_losses))
````

````python
import numpy as np
import time

# 1. Create dataset
arr = np.random.rand(500, 100)

print("Array shape:", arr.shape)

# -----------------------------------------
# (a) Python loop approach
# -----------------------------------------
start_a = time.perf_counter()

res_a = [np.sqrt((row**2).sum()) for row in arr]

end_a = time.perf_counter()
time_a = end_a - start_a

print("\n(a) Loop time:", time_a)

# -----------------------------------------
# (b) Fully vectorized approach
# -----------------------------------------
start_b = time.perf_counter()

res_b = np.sqrt((arr**2).sum(axis=1))

end_b = time.perf_counter()
time_b = end_b - start_b

print("(b) Vectorized time:", time_b)

# -----------------------------------------
# (c) In-place vectorized approach
# -----------------------------------------
start_c = time.perf_counter()

tmp = arr.copy()
tmp **= 2
res_c = np.sqrt(tmp.sum(axis=1))

end_c = time.perf_counter()
time_c = end_c - start_c

print("(c) In-place vectorized time:", time_c)

# -----------------------------------------
# 4. Speedup calculation
# -----------------------------------------
speedup_b_over_a = time_a / time_b if time_b > 0 else np.inf

print("\n=== RESULTS ===")
print("Time (a) loop:", time_a)
print("Time (b) vectorized:", time_b)
print("Time (c) in-place vectorized:", time_c)

print("\nSpeedup (b over a):", speedup_b_over_a)

# -----------------------------------------
# 5. Optional correctness check
# -----------------------------------------
print("\nConsistency check:")
print("a vs b:", np.allclose(res_a, res_b))
print("b vs c:", np.allclose(res_b, res_c))
````

**Q100.**  
**Full ML pipeline challenge** — implement everything from scratch using only Sections 1–13:

```python
X = np.random.rand(200, 5)
y = ((X[:, 0] + X[:, 1]) > 1.0).astype(int)  # binary target
```

1. **Z-score standardize** `X` column-wise using broadcasting
2. **Shuffle and split** 70/15/15 using `np.argsort(np.random.rand(200))` as shuffle index
3. **Initialize weights** `w = np.zeros(5)` and bias `b = 0.0`
4. **Forward pass** on `X_train`: `z = X_train @ w + b`, `y_hat = 1 / (1 + np.exp(-z))`
5. **BCE loss**: clip `y_hat` to `[1e-7, 1-1e-7]`, compute `-np.mean(y*log + (1-y)*log)`
6. **Gradient**: `grad_w = X_train.T @ (y_hat - y_train) / len(y_train)`
7. **One gradient descent step**: `w -= 0.1 * grad_w`
8. **Forward pass again** with updated `w`, compute new BCE loss
9. Print: initial loss, updated loss (should be slightly lower), gradient norm `np.sqrt((grad_w**2).sum())`

---

````python
import numpy as np

# 1. Data
X = np.random.rand(200, 5)
y = ((X[:, 0] + X[:, 1]) > 1.0).astype(int)

print("X shape:", X.shape)
print("y distribution:", np.bincount(y))

# -----------------------------------------
# 2. Z-score standardization (column-wise)
# -----------------------------------------
mean = X.mean(axis=0)
std = X.std(axis=0)
X = (X - mean) / std

print("\nAfter standardization (first 2 rows):")
print(X[:2])

# -----------------------------------------
# 3. Shuffle and split (70/15/15)
# -----------------------------------------
idx = np.argsort(np.random.rand(200))

X = X[idx]
y = y[idx]

n = len(X)
n_train = int(0.7 * n)
n_val = int(0.15 * n)

X_train = X[:n_train]
y_train = y[:n_train]

X_val = X[n_train:n_train + n_val]
y_val = y[n_train:n_train + n_val]

X_test = X[n_train + n_val:]
y_test = y[n_train + n_val:]

print("\nSplit sizes:")
print("Train:", X_train.shape)
print("Val:", X_val.shape)
print("Test:", X_test.shape)

# -----------------------------------------
# 4. Initialize parameters
# -----------------------------------------
w = np.zeros(5)
b = 0.0

# -----------------------------------------
# Helper: sigmoid
# -----------------------------------------
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# -----------------------------------------
# 5. Forward pass
# -----------------------------------------
z = X_train @ w + b
y_hat = sigmoid(z)

# -----------------------------------------
# 6. BCE loss (initial)
# -----------------------------------------
eps = 1e-7
y_hat_clipped = np.clip(y_hat, eps, 1 - eps)

loss1 = -np.mean(
    y_train * np.log(y_hat_clipped) +
    (1 - y_train) * np.log(1 - y_hat_clipped)
)

print("\nInitial loss:", loss1)

# -----------------------------------------
# 7. Gradient computation
# -----------------------------------------
grad_w = (X_train.T @ (y_hat - y_train)) / len(y_train)

grad_norm = np.sqrt((grad_w ** 2).sum())

print("Gradient norm:", grad_norm)

# -----------------------------------------
# 8. One gradient descent step
# -----------------------------------------
w = w - 0.1 * grad_w

# -----------------------------------------
# 9. Forward pass after update
# -----------------------------------------
z2 = X_train @ w + b
y_hat2 = sigmoid(z2)

y_hat2_clipped = np.clip(y_hat2, eps, 1 - eps)

loss2 = -np.mean(
    y_train * np.log(y_hat2_clipped) +
    (1 - y_train) * np.log(1 - y_hat2_clipped)
)

print("\nUpdated loss:", loss2)

# -----------------------------------------
# 10. Final comparison
# -----------------------------------------
print("\n=== SUMMARY ===")
print("Initial loss :", loss1)
print("Updated loss :", loss2)
print("Loss decreased?:", loss2 < loss1)
print("Gradient norm:", grad_norm)
````

A retail company has recorded weekly sales (in ₹ thousands) for 4 stores over 12 weeks. The data arrived as a flat 1D array of 48 values with some entries corrupted as np.nan. Your task is to restructure and analyze it.

Reshape the 1D array into a (4, 12) matrix — rows = stores, columns = weeks.
Replace all NaN values with the column (weekly) mean across non-NaN stores.

````python
import numpy as np

#  random generator
rng = np.random.default_rng(101)

# Generate sales data (mean=65, std=15)
sales_data = rng.normal(loc=65, scale=15, size=48) 
# loc=65 means average value, scale=15 means variation (standard deviation)



# Round and keep positive values
sales_data = np.round(np.clip(sales_data, 20, None), 1)
 # Clip values to ensure minimum sales is 20 (no negative or too small values)


#  missing values
missing_idx = rng.choice(48, size=6, replace=False) # Randomly select 6 positions from 48 values to simulate missing data



sales_data[missing_idx] = np.nan # Replace selected positions with NaN (missing values)



print("Original 1D Data:")
print(sales_data)

````
