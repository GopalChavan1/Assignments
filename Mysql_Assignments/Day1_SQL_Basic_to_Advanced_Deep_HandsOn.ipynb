# 3. SQL Query Execution Order

A simplified logical processing order is:

```text
FROM
  ↓
JOIN
  ↓
WHERE
  ↓
GROUP BY
  ↓
HAVING
  ↓
SELECT
  ↓
DISTINCT
  ↓
ORDER BY
  ↓
LIMIT
```

This explains why you normally cannot use an aggregate filter in `WHERE`.

Incorrect:

```sql
WHERE COUNT(*) > 5
```

Correct:

```sql
GROUP BY department_id
HAVING COUNT(*) > 5
```

### Important distinction

`WHERE` → filters individual rows.

`HAVING` → filters groups after aggregation.

# 7. NULL — A Very Important SQL Concept

`NULL` means missing/unknown/not available.

This is **not** the same as:

- 0
- empty string
- false

Wrong:

```sql
WHERE manager_id = NULL
```

Correct:

```sql
WHERE manager_id IS NULL
```

Not null:

```sql
WHERE manager_id IS NOT NULL
```


````python
SELECT employee_id, employee_name, manager_id
FROM employees
WHERE manager_id IS NULL;
````

````python
SELECT employee_id, employee_name, manager_id
FROM employees
WHERE manager_id IS NOT NULL
LIMIT 20;
````

# 16. WHERE vs HAVING — Deep Practice

Consider:

```sql
SELECT department_id, AVG(salary)
FROM employees
WHERE status = 'Active'
GROUP BY department_id
HAVING AVG(salary) > 90000;
```

Logical meaning:

1. Start with employees.
2. Keep active employees.
3. Group them by department.
4. Calculate average salary.
5. Keep departments whose average is above 90,000.

### Rule

If the condition applies to **individual rows**, think `WHERE`.

If it applies to an **aggregate/group**, think `HAVING`.


### Practice 14

Find active departments having at least 5 active employees and average salary above 80,000.

**Hint:** WHERE filters active rows. HAVING handles COUNT and AVG.

**Solution:**


````python
SELECT department_id,
       COUNT(*) AS active_employees,
       ROUND(AVG(salary), 2) AS avg_salary
FROM employees
WHERE status = 'Active'
GROUP BY department_id
HAVING COUNT(*) >= 5
   AND AVG(salary) > 80000
ORDER BY avg_salary DESC;
````

# 18. Correlated Subquery

A correlated subquery references a row from the outer query.

Business question:

> Which employees earn more than the average salary of their own department?

The inner query must calculate the average for the current employee's department.


````python
SELECT e.employee_id,
       e.employee_name,
       e.department_id,
       e.salary
FROM employees e
WHERE e.salary > (
    SELECT AVG(e2.salary)
    FROM employees e2
    WHERE e2.department_id = e.department_id
)
ORDER BY e.department_id, e.salary DESC;
````

### Practice 16

# 20. CTE — Common Table Expression

CTEs make multi-step SQL easier to understand.

Syntax:

```sql
WITH cte_name AS (
    SELECT ...
)
SELECT ...
FROM cte_name;
```

Example:


````python
WITH department_stats AS (
    SELECT department_id,
           COUNT(*) AS employee_count,
           AVG(salary) AS avg_salary
    FROM employees
    GROUP BY department_id
)
SELECT *
FROM department_stats
WHERE avg_salary > 90000
ORDER BY avg_salary DESC;
````

### Why use a CTE?

- readability
- reusable intermediate result
- easier debugging
- complex analytical logic
- recursive queries later

A CTE is not automatically faster than a subquery; use it primarily for clarity unless execution behavior has been verified.


### Practice 18

Using a CTE, calculate department payroll and return departments with payroll above 1,000,000.

**Hint:** SUM(salary), GROUP BY, then filter in outer query.

**Solution:**


````python
WITH dept_payroll AS (
    SELECT department_id,
           SUM(salary) AS payroll
    FROM employees
    GROUP BY department_id
)
SELECT department_id, payroll
FROM dept_payroll
WHERE payroll > 1000000
ORDER BY payroll DESC;
````

# 26. Advanced Day-1 Challenge — Above Department Average

### Question 30

Return:

- employee_id
- employee_name
- department_id
- salary
- department average salary
- difference from department average

**Hint:** Start with a department aggregate/CTE or correlated subquery.

### Solution


````python
WITH dept_avg AS (
    SELECT department_id,
           AVG(salary) AS avg_salary
    FROM employees
    GROUP BY department_id
)
SELECT e.employee_id,
       e.employee_name,
       e.department_id,
       e.salary,
       ROUND(d.avg_salary, 2) AS department_avg_salary,
       ROUND(e.salary - d.avg_salary, 2) AS difference
FROM employees e
JOIN dept_avg d
  ON d.department_id = e.department_id
WHERE e.salary > d.avg_salary
ORDER BY difference DESC;
````

# 29. Query Debugging Checklist

When a SQL result is wrong, ask:

### 1. What is the expected row grain?
One employee? One department? One order?

### 2. Are joins creating extra rows?
Check counts before and after each join.

### 3. Is NULL affecting the logic?
Use `IS NULL`, `IS NOT NULL`, `COALESCE`.

### 4. Is filtering happening at the correct stage?
Row filter → WHERE  
Group filter → HAVING

### 5. Is aggregation correct?
Check GROUP BY columns.

### 6. Did DISTINCT hide a real data problem?
Do not use DISTINCT as a blind duplicate fix.

### 7. Is the date filter correct?
Be careful with date vs datetime boundaries.

### 8. Does the query answer the business question?
A syntactically valid query can still be logically wrong.


# 30. Performance Awareness — Day 1 Preview

# 31. 3–4 Year Experience Interview Questions

### Q1. WHERE vs HAVING?
**Answer:** WHERE filters rows before grouping; HAVING filters groups after aggregation.

### Q2. COUNT(*) vs COUNT(column)?
**Answer:** COUNT(*) counts rows; COUNT(column) ignores NULL values in that column.

### Q3. Why is `= NULL` incorrect?
**Answer:** NULL represents unknown/missing and uses three-valued logic. Use `IS NULL`.

### Q4. What is a correlated subquery?
**Answer:** A subquery that references a column from the outer query and is logically evaluated with respect to outer rows.

### Q5. UNION vs UNION ALL?
**Answer:** UNION removes duplicate result rows; UNION ALL preserves them and generally avoids the duplicate-elimination step.

### Q6. What is query grain?
**Answer:** The meaning represented by one output row. Defining grain helps prevent incorrect joins and aggregation.

### Q7. Can DISTINCT fix a bad JOIN?
**Answer:** It may hide duplicate output, but it does not correct an incorrect relationship or aggregation logic.

### Q8. Why use CTEs?
**Answer:** Mainly readability, modularity and multi-step logic. Performance should be validated rather than assumed.

### Q9. What is a scalar subquery?
**Answer:** A subquery expected to return a single value.

### Q10. What happens if a scalar subquery returns multiple rows?
**Answer:** In MySQL, the scalar-subquery context raises an error.

### Q11. What is the difference between IN and EXISTS?
**Answer:** Both can express membership/existence logic, but their optimization behavior depends on the query and data. EXISTS is naturally expressed as a boolean existence test.

### Q12. What is SARGability?
**Answer:** Writing predicates so the optimizer can efficiently use suitable indexes, rather than applying functions/transforms that prevent efficient index access in relevant cases.

### 1–10


````python
-- 1
SELECT * FROM employees;

-- 2
SELECT employee_name, salary FROM employees;

-- 3
SELECT employee_name, salary
FROM employees
WHERE salary > 100000;

-- 4
SELECT employee_id, employee_name
FROM employees
WHERE status = 'Active';

-- 5
SELECT employee_id, employee_name, department_id
FROM employees
WHERE department_id IN (1,2,3);

-- 6
SELECT employee_id, employee_name
FROM employees
WHERE manager_id IS NULL;

-- 7
SELECT employee_name, salary
FROM employees
ORDER BY salary DESC
LIMIT 10;

-- 8
SELECT employee_name, salary*12 AS annual_salary
FROM employees;

-- 9
SELECT employee_name, salary,
       CASE
         WHEN salary >= 150000 THEN 'Executive'
         WHEN salary >= 100000 THEN 'Senior'
         WHEN salary >= 70000 THEN 'Mid'
         ELSE 'Junior'
       END AS salary_band
FROM employees;

-- 10
SELECT COUNT(*) AS employee_count
FROM employees;
````

### 11–20


````python
-- 11
SELECT department_id, COUNT(*) employee_count
FROM employees
GROUP BY department_id;

-- 12
SELECT department_id, ROUND(AVG(salary),2) avg_salary
FROM employees
GROUP BY department_id;

-- 13
SELECT department_id, COUNT(*) employee_count
FROM employees
GROUP BY department_id
HAVING COUNT(*) > 5;

-- 14
SELECT department_id, AVG(salary) avg_salary
FROM employees
GROUP BY department_id
HAVING AVG(salary) > 90000;

-- 15
SELECT segment, COUNT(*) customer_count
FROM customers
GROUP BY segment;

-- 16
SELECT city, COUNT(*) customer_count
FROM customers
GROUP BY city
ORDER BY customer_count DESC;

-- 17
SELECT sales_channel, COUNT(*) order_count, SUM(total_amount) revenue
FROM orders
GROUP BY sales_channel;

-- 18
SELECT customer_id, SUM(total_amount) total_value
FROM orders
GROUP BY customer_id
HAVING SUM(total_amount) > 100000;

-- 19
SELECT employee_id, employee_name, salary
FROM employees
WHERE salary > (SELECT AVG(salary) FROM employees);

-- 20
SELECT e.employee_id, e.employee_name, e.department_id, e.salary
FROM employees e
WHERE e.salary > (
    SELECT AVG(e2.salary)
    FROM employees e2
    WHERE e2.department_id=e.department_id
);
````

### 21–30


````python
-- 21
SELECT MAX(salary) AS second_highest
FROM employees
WHERE salary < (SELECT MAX(salary) FROM employees);

-- 22
SELECT e.employee_id, e.employee_name
FROM employees e
WHERE NOT EXISTS (
    SELECT 1
    FROM performance_reviews p
    WHERE p.employee_id=e.employee_id
);

-- 23
SELECT c.customer_id, c.customer_name
FROM customers c
WHERE NOT EXISTS (
    SELECT 1
    FROM orders o
    WHERE o.customer_id=c.customer_id
);

-- 24
SELECT department_id,
       COUNT(*) active_count,
       AVG(salary) avg_salary
FROM employees
WHERE status='Active'
GROUP BY department_id
HAVING COUNT(*) > 5
   AND AVG(salary) > 80000;

-- 25
WITH dept_payroll AS (
    SELECT department_id, SUM(salary) payroll
    FROM employees
    GROUP BY department_id
)
SELECT *
FROM dept_payroll
ORDER BY payroll DESC;

-- 26
SELECT e.employee_id, e.employee_name, e.department_id, e.salary
FROM employees e
WHERE e.salary = (
    SELECT MAX(e2.salary)
    FROM employees e2
    WHERE e2.department_id=e.department_id
);

-- 27
SELECT customer_email, COUNT(*) duplicate_count
FROM customers
GROUP BY customer_email
HAVING COUNT(*) > 1;

-- 28
-- Rewrite any query from the day using a subquery and then a CTE.
-- Compare readability and result equivalence.

-- 29
EXPLAIN
SELECT employee_id, employee_name
FROM employees
WHERE department_id=1
  AND salary>=100000;

-- 30
-- Write one sentence before each query:
-- "One output row represents ________." 
````
