### NP17. Normalize data using NumPy (Min-Max scaling).


```python
# NP17. Normalize data using NumPy (Min-Max scaling).


# Create a DataFrame with feature_1=[10,20,30,40,50], feature_2=[100,200,300,400,500]
# Apply Min-Max normalization (range [0, 1]) to both features

import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    'feature_1': [10, 20, 30, 40, 50],
    'feature_2': [100, 200, 300, 400, 500]
})

# Extract array
arr = df[['feature_1', 'feature_2']].values

# Min-Max normalization
normalized = (arr - arr.min(axis=0)) / (arr.max(axis=0) - arr.min(axis=0))

# Wrap back in DataFrame
df_normalized = pd.DataFrame(normalized, columns=df[['feature_1', 'feature_2']].columns)

print(df_normalized)
```

### NP18. Standardize data using Z-score with NumPy.


```python
# NP18. Standardize data using Z-score with NumPy.

# Create a DataFrame with feature_1=[10,20,30,40,50], feature_2=[100,200,300,400,500]
# Apply Z-score standardization (mean=0, std=1) to both features

import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    'feature_1': [10, 20, 30, 40, 50],
    'feature_2': [100, 200, 300, 400, 500]
})

# Extract array
arr = df[['feature_1', 'feature_2']].values

# Z-score standardization
z = (arr - arr.mean(axis=0)) / arr.std(axis=0)

# Wrap back in DataFrame
df_standardized = pd.DataFrame(z, columns=df[['feature_1', 'feature_2']].columns)

print(df_standardized)
```

### NP42. Detect and replace outliers using NumPy.


```python
# NP42. Detect and replace outliers using NumPy.
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({'values': [1, 2, 3, 4, 5, 100, 7, 8, 9, 10]})

# Compute quartiles
q1 = np.percentile(df['values'], 25)
q3 = np.percentile(df['values'], 75)
iqr = q3 - q1

# Define bounds
lo = q1 - 1.5 * iqr
hi = q3 + 1.5 * iqr

# Clip outliers
df['clipped'] = np.clip(df['values'], lo, hi)

print(f"Q1: {q1}, Q3: {q3}, IQR: {iqr}")
print(f"Bounds: [{lo}, {hi}]")
print(df)
```

```python
# NP43. Scale dataset using NumPy functions.
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    'feature_1': [10, 20, 30, 40, 50],
    'feature_2': [100, 200, 300, 400, 500]
})

# Extract array
arr = df[['feature_1', 'feature_2']].values

# Min-Max scaling
df['scaled_1'] = (arr[:, 0] - arr[:, 0].min()) / (arr[:, 0].max() - arr[:, 0].min())
df['scaled_2'] = (arr[:, 1] - arr[:, 1].min()) / (arr[:, 1].max() - arr[:, 1].min())

print(df)
```

```python
# NP44. Engineer new features using NumPy.
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({'value': [1, 2, 3, 4, 5]})

# Polynomial features
df['value_squared'] = np.square(df['value'])
df['value_power_3'] = np.power(df['value'], 3)

# Log feature
df['log_value'] = np.log1p(df['value'])

# Interaction feature
df['interaction'] = df['value'] * df['value_squared']

print(df)
```

### NP46. Compute differences using `np.diff`.


```python
# NP46. Compute differences using `np.diff`.
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({'values': [1, 3, 6, 10, 15]})

# Compute differences
diff = np.diff(df['values'].values)

# Prepend NaN
diff_with_nan = np.insert(diff, 0, np.nan)

df['values_diff'] = diff_with_nan

print(df)
```

### NP47. Apply conditional encoding using NumPy.


```python
# NP47. Apply conditional encoding using NumPy.
import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({'category': ['DS', 'ML', 'AI', 'DS', 'ML', 'AI']})

# Encode using np.select
encoded = np.select(
    [df['category'] == 'DS', df['category'] == 'ML', df['category'] == 'AI'],
    [0, 1, 2],
    default=-1
)

df['encoded'] = encoded

print(df)
```

### NP49. Perform a mini data analysis project.


```python
# NP49. Perform a mini data analysis project.
import pandas as pd
import numpy as np

# 1. Generate data with NumPy
np.random.seed(42)
df = pd.DataFrame({
    'feature_1': np.random.normal(50, 10, 100),
    'feature_2': np.random.normal(100, 20, 100),
    'category': np.random.choice(['A', 'B', 'C'], 100)
})

# 2. Clean with Pandas
df = df.dropna()

# 3. Engineer features with NumPy
df['log_feature_1'] = np.log1p(df['feature_1'])
df['interaction'] = df['feature_1'] * df['feature_2']

# 4. GroupBy analysis
group_stats = df.groupby('category').agg({
    'feature_1': ['mean', 'std'],
    'feature_2': ['mean', 'std']
})

# 5. Detect outliers
q1 = np.percentile(df['feature_1'], 25)
q3 = np.percentile(df['feature_1'], 75)
iqr = q3 - q1
outliers = df[(df['feature_1'] < q1 - 1.5 * iqr) | (df['feature_1'] > q3 + 1.5 * iqr)]

# 6. Export summary
summary = df.groupby('category').mean()

print(f"Dataset shape: {df.shape}")
print(f"Outliers found: {outliers.shape[0]}")
print("\nSummary by category:")
print(summary)
```

### NP50. Build an end-to-end pipeline using NumPy + Pandas.


```python
# NP50. Build an end-to-end pipeline using NumPy + Pandas.
import pandas as pd
import numpy as np

# 1. Generate raw data
np.random.seed(42)
n = 1000

raw_data = np.random.normal(50, 15, n)
raw_data = np.clip(raw_data, 0, 100)  # Clean invalid values

df = pd.DataFrame({
    'feature_1': raw_data,
    'feature_2': np.random.normal(100, 25, n),
    'feature_3': np.random.randint(0, 10, n)
})

# 2. Feature engineering
df['log_f1'] = np.log1p(df['feature_1'])
df['square_f1'] = np.square(df['feature_1'])
df['interaction'] = df['feature_1'] * df['feature_2']
df['label'] = np.where(df['feature_1'] + df['feature_2'] > 150, 1, 0)

# 3. Statistical analysis
group_stats = df.groupby('label').agg({
    'feature_1': ['mean', 'std'],
    'feature_2': ['mean', 'std']
})

# 4. Outlier removal
q1 = np.percentile(df['feature_1'], 25)
q3 = np.percentile(df['feature_1'], 75)
iqr = q3 - q1
df['feature_1_clean'] = np.clip(df['feature_1'], q1 - 1.5 * iqr, q3 + 1.5 * iqr)

# 5. ML feature matrix
features = ['feature_1_clean', 'feature_2', 'feature_3', 'log_f1', 'square_f1', 'interaction']
X = df[features].to_numpy(dtype=np.float32)
y = df['label'].to_numpy()

# 6. Export
df.to_csv('ml_dataset.csv', index=False)

print(f"Pipeline complete!")
print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
print(f"Dataset saved to: ml_dataset.csv")
```
