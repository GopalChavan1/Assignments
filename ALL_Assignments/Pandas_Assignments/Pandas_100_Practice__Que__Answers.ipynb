```python
def q4():
    s = pd.Series([1,2,3,4,5])
    addition = s + 10 
    subtraction = s - 1 
    multiplication = s * 2
    division = s / 2
    power = s ** 2
    return {'add': addition, 'sub': subtraction, 'mul': multiplication, 'div': division, 'pow': power}
print(q4())
```

### Q5. Find mean, min, max of a Series.


```python
def q5():
    s = pd.Series([10, 20, 30, 40, 50])
    mean_val = s.mean()
    min_val = s.min()
    max_val = s.max()
    sum_val = s.sum()
    median_val = s.median()
    std_val = s.std()
    return {'mean': mean_val, 'min': min_val, 'max': max_val, 'sum': sum_val, 'median': median_val, 'std': std_val}
print(q5())
```

### Q56. Detect outliers in a column.


```python
# Q56. Detect outliers in a column.
def q56():
    df = pd.DataFrame({'num': [1, 2, 3, 4, 5, 100]})
    q1 = df['num'].quantile(0.25)
    q3 = df['num'].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = df[(df['num'] < lower_bound) | (df['num'] > upper_bound)]
    return {'q1': q1, 'q3': q3, 'iqr': iqr, 'lower': lower_bound, 'upper': upper_bound, 'outliers': outliers}
```

### Q57. Handle outliers using clipping.


```python
# Q57. Handle outliers using clipping.
def q57():
    df = pd.DataFrame({'num': [1, 2, 3, 4, 5, 100]})
    q1 = df['num'].quantile(0.25)
    q3 = df['num'].quantile(0.75)
    iqr = q3 - q1
    lo = q1 - 1.5 * iqr
    hi = q3 + 1.5 * iqr
    clipped = df['num'].clip(lower=lo, upper=hi)
    return clipped
```

## 1. groupby()

Used to split data into groups and perform aggregations.

Very flexible.

Returns a flat structure (Series/DataFrame).

Can aggregate multiple columns and apply custom functions.

```python
import pandas as pd

df = pd.DataFrame({
    'Department': ['IT', 'IT', 'HR', 'HR'],
    'Salary': [50000, 60000, 40000, 45000]
})

result = df.groupby('Department')['Salary'].mean()

print(result)
```

## 2. pivot()

Reshapes data into a 2D table.

Does NOT perform aggregation.

Requires unique combinations of index and columns.

Fails if duplicates exist.

```python
import pandas as pd

df = pd.DataFrame({
    'Date': ['2025-01', '2025-02'],
    'Product': ['A', 'A'],
    'Sales': [100, 150]
})

result = df.pivot(
    index='Date',
    columns='Product',
    values='Sales'
)

print(result)
```

## 3. pivot_table()

Similar to pivot().

Handles duplicate values.

Performs aggregation automatically.

Default aggregation function is mean.

```python
import pandas as pd

df = pd.DataFrame({
    'Date': ['2025-01', '2025-01'],
    'Product': ['A', 'A'],
    'Sales': [100, 150]
})

result = pd.pivot_table(
    df,
    index='Date',
    columns='Product',
    values='Sales',
    aggfunc='mean'
)

print(result)
```

groupby() is used for flexible grouping and aggregation, 

pivot() reshapes data without aggregation and fails on duplicates,

 while pivot_table() reshapes data and handles duplicates through aggregation.
