```python
# Q78. Distance from One Point to Multiple Lines

import numpy as np

P0 = np.array([[0,0],[1,1]])
P1 = np.array([[2,0],[3,1]])

p = np.array([1,2])

line_vec = P1 - P0
point_vec = p - P0

dist = np.abs(
    line_vec[:,0]*point_vec[:,1] -
    line_vec[:,1]*point_vec[:,0]
) / np.linalg.norm(line_vec, axis=1)

print(dist)
```

```python
# Q79. Distance from Many Points to Many Lines

import numpy as np 

P0 = np.array([[0,0],[1,1]])
P1 = np.array([[2,0],[3,1]])

P = np.array([
    [1,2],
    [2,3],
    [3,4]
])

line_vec = P1 - P0

distances = []

for p in P:
    point_vec = p - P0

    dist = np.abs(
        line_vec[:,0]*point_vec[:,1] -
        line_vec[:,1]*point_vec[:,0]
    ) / np.linalg.norm(line_vec, axis=1)

    distances.append(dist)

print(np.array(distances))

```

```python
# Q97. Einsum Equivalent of Common Operations

import numpy as np

A = np.array([1,2,3])
B = np.array([4,5,6])

# Inner product
inner = np.einsum('i,i->', A, B)

# Outer product
outer = np.einsum('i,j->ij', A, B)

# Sum
total = np.einsum('i->', A)

# Elementwise multiply
mul = np.einsum('i,i->i', A, B)

print("Inner =", inner)
print("Outer =\n", outer)
print("Sum =", total)
print("Multiply =", mul)
```

```python
# Q98. Sample a Path Using Equidistant Points

import numpy as np

X = np.array([0,1,2,4])
Y = np.array([0,1,0,0])

distance = np.cumsum(
    np.sqrt(np.diff(X)**2 + np.diff(Y)**2)
)

distance = np.insert(distance, 0, 0)

new_dist = np.linspace(0, distance[-1], 10)

new_X = np.interp(new_dist, distance, X)
new_Y = np.interp(new_dist, distance, Y)

print(new_X)
print(new_Y)
```

```python
# Q99. Find Rows That Form a Multinomial Distribution

import numpy as np

n = 5

X = np.array([
    [2,2,1],
    [1,1,1],
    [3,1,1],
    [2.5,1,1]
])

mask = (
    (np.mod(X,1) == 0).all(axis=1)
    &
    (X.sum(axis=1) == n)
)

print(X[mask])
```

```python
# Q100. Bootstrap 95% Confidence Interval

import numpy as np

X = np.array([10,12,15,18,20])

N = 10000

means = []

for _ in range(N):

    sample = np.random.choice(
        X,
        size=len(X),
        replace=True
    )

    means.append(sample.mean())

means = np.array(means)

lower = np.percentile(means, 2.5)
upper = np.percentile(means, 97.5)

print("95% Confidence Interval:")
print(lower, upper)
```
