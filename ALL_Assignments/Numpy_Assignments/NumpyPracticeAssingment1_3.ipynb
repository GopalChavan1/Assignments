###  Q2: Print square of each number using loop


````python
# Q2: Print square of each number using loop
numbers = [1, 2, 3, 4, 5]

print("Squares using loop:")
for num in numbers:
    print(num ** 2)
````

````python
# Using list comprehension
squares = [num ** 2 for num in numbers]
print("\nSquares using list comprehension:")
print(squares)
````

### **Q3.** Demonstrate the **view vs copy** difference:  
 Create `original = np.array([10, 20, 30, 40, 50])`.  
 - Make a **view** of the last 3 elements. Change the view's first value to `0`. Print `original`.  
 - Make a **copy** of the first 3 elements. Change the copy's first value to `0`. Print `original` again.  
 - Explain what you observe.

````python
import numpy as np
````

````python
original = np.array([10, 20, 30, 40, 50])
````

````python
print("Original Array:")
print(original)
````

````python
# Create View
view_arr = original[2:] #A view shares memory with the original array.

view_arr[0] = 0
````

````python
print("\nAfter modifying View:")
print("View:", view_arr)
print("Original:", original)
````

````python
# Create Copy
copy_arr = original[:3].copy() #A copy creates a completely new array.
copy_arr[0] = 0
````

````python
print("\nAfter modifying Copy:")
print("Copy:", copy_arr)
print("Original:", original)
````
