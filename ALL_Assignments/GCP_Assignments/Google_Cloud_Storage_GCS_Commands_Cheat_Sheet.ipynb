# Google Cloud Storage (GCS) Commands Cheat Sheet
Most important `gcloud storage` commands.


## Check Current Project


```python
gcloud config get-value project
```

## List Projects


```python
gcloud projects list
```

## Set Project


```python
gcloud config set project PROJECT_ID
```

## List Buckets


```python
gcloud storage buckets list
```

## Create Bucket


```python
gcloud storage buckets create gs://bucket-name
```

## Create Bucket Region


```python
gcloud storage buckets create gs://bucket-name --location=us-central1
```

## Create Bucket Storage Class


```python
gcloud storage buckets create gs://bucket-name --default-storage-class=STANDARD
```

## Describe Bucket


```python
gcloud storage buckets describe gs://bucket-name
```

## Delete Bucket


```python
gcloud storage buckets delete gs://bucket-name
```

## List Objects


```python
gcloud storage ls gs://bucket-name
```

## Upload File


```python
gcloud storage cp file.txt gs://bucket-name
```

## Upload Folder


```python
gcloud storage cp --recursive folder gs://bucket-name
```

## Download File


```python
gcloud storage cp gs://bucket-name/file.txt .
```

## Download Folder


```python
gcloud storage cp --recursive gs://bucket-name/folder .
```

## Copy File


```python
gcloud storage cp gs://bucket1/file.txt gs://bucket2/
```

## Move/Rename


```python
gcloud storage mv gs://bucket/old.txt gs://bucket/new.txt
```

## Delete File


```python
gcloud storage rm gs://bucket/file.txt
```

## Delete Folder


```python
gcloud storage rm --recursive gs://bucket/folder
```

## View IAM


```python
gcloud storage buckets get-iam-policy gs://bucket-name
```

## Add IAM


```python
gcloud storage buckets add-iam-policy-binding gs://bucket-name --member=user:user@gmail.com --role=roles/storage.objectViewer
```

## Remove IAM


```python
gcloud storage buckets remove-iam-policy-binding gs://bucket-name --member=user:user@gmail.com --role=roles/storage.objectViewer
```

## Enable Versioning


```python
gcloud storage buckets update gs://bucket-name --versioning
```

## Disable Versioning


```python
gcloud storage buckets update gs://bucket-name --no-versioning
```

## Add Label


```python
gcloud storage buckets update gs://bucket-name --update-labels=env=dev
```

## Remove Label


```python
gcloud storage buckets update gs://bucket-name --remove-labels=env
```

## Sync Local->Bucket


```python
gcloud storage rsync folder gs://bucket-name
```

## Sync Bucket->Local


```python
gcloud storage rsync gs://bucket-name folder
```

## Legacy gsutil List


```python
gsutil ls
```

## Legacy Upload


```python
gsutil cp file.txt gs://bucket
```

## Legacy Download


```python
gsutil cp gs://bucket/file.txt .
```
